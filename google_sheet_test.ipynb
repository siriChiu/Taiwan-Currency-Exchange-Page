```python
import base64
import json
import gspread
from oauth2client.service_account import ServiceAccountCredentials
import configparser

# Define the scope
scope = ["https://spreadsheets.google.com/feeds", "https://www.googleapis.com/auth/drive"]

#  Load config .ini file
cfg_file = "./config.ini"
with open(cfg_file, 'r') as f:
    cfg = f.read()
# Parse .ini file
config = configparser.ConfigParser()
config.read_string(cfg)
cfg_dict = {section: dict(config.items(section)) for section in config.sections()}

if 'user_info' not in cfg_dict:
    print('Missing user_info section in the config file')
    exit()

owner = cfg_dict['user_info']['owner']
company = cfg_dict['user_info']['company']
# Load credentials from the JSON key file
crededntial_file = "./credential"
with open(crededntial_file, 'r') as f:
    credential_coded = f.read()

credential = base64.b64decode(credential_coded).decode('utf-8')
credential_dict = json.loads(credential)

creds = ServiceAccountCredentials.from_json_keyfile_dict(credential_dict, scope)
client = gspread.authorize(creds)

# Open the Google Sheet
spreadsheet_url_coded = "aHR0cHM6Ly9kb2NzLmdvb2dsZS5jb20vc3ByZWFkc2hlZXRzL2QvMTVxUEJ0Wl9xWkN3WDNCdHJPandZMkhxRjBPOHktRTBURlBYNGozZXlmYkEvZWRpdD91c3A9c2hhcmluZw=="
spreadsheet_url = base64.b64decode(spreadsheet_url_coded).decode('utf-8')

try:
    sheet = client.open_by_url(spreadsheet_url).sheet1
except gspread.exceptions.APIError:
    print('Please share the Google Sheet with the client email in the credential.json file')

# Read a column (e.g., Column A)
column_data = sheet.get_all_values()  # 1 means column A
companies = sheet.col_values(1)
owners = sheet.col_values(2)

company_idx = [i for i, x in enumerate(companies) if x == company]
owner_idx = [i for i, x in enumerate(owners) if x == owner]
if len(company_idx) == 0 or len(owner_idx) == 0:
    print('Not found')
    exit()
elif len(company_idx) > 1 or len(owner_idx) > 1:
    print('Multiple found')
    exit()
elif company_idx[0] != owner_idx[0]:
    print('Not match')
    exit()

column_header = column_data[0]

company_owner_idx = company_idx[0]
company_owner_info = column_data[company_owner_idx]

for i in range(len(column_header)):
    print(f'{column_header[i]}: {company_owner_info[i]}')

# find Status index
status_idx = column_header.index('Status')
status = company_owner_info[status_idx]
print(f'Status: {status}')
```
